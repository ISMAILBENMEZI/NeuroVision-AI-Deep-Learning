```python
from pathlib import Path
import zipfile
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

# from google.colab import drive
# drive.mount('/content/drive')

# data_path = Path(
#     "/content/drive/MyDrive/DataSets/NeuroVision-AI-Deep-Learning/data/raw"
# )

# with zipfile.ZipFile("/content/drive/MyDrive/DataSets/data.zip") as zipf:
#     zipf.extractall("/content")

data_path = Path("/kaggle/input/datasets/ismailbenmezi/datasetneurovesion/Data")
```

# Image Resizing and Array Formatting

```python
image_extensions = {".jpg" , ".jpeg" , ".png" , ".bmp"}
class_names = ["glioma", "meningioma", "notumor", "pituitary"]

image_size = (224, 224)

images = []
labels = []

for label , class_name in enumerate(class_names):
    class_path = data_path / class_name
    for image_path in class_path.iterdir():
        if image_path.suffix.lower() in image_extensions:
            image = cv2.imread(str(image_path))
            if image is not None:
                image = cv2.cvtColor(image , cv2.COLOR_BGR2RGB)
                image = cv2.resize(image , image_size)

                images.append(image)
                labels.append(label)

# display(images)
# display(labels)

X = np.array(images , dtype=np.uint8)
y = np.array(labels)

# print(X)

print("Images shape:", X.shape)
print("Labels shape:", y.shape)
print("Classes:", dict(enumerate(class_names)))
```

# Normalization

```python
# # X = X.astype(np.) / 255.0
X = cv2.normalize(X , None , alpha=0.0 , beta=1.0 , norm_type=cv2.NORM_MINMAX, dtype=cv2.CV_32F)
print("Images shape:", X.shape)
print("Minimum pixel value:", X.min())
print("Maximum pixel value:", X.max())
print("Data type:", X.dtype)




# X = X.astype(np.float32) / 255.0
# image = []
# for x in X:
#     image.append(x.astype(np.int8) / 255)



# print("Images shape:", X.shape)
# print("Minimum pixel value:", X.min())
# print("Maximum pixel value:", X.max())
# print("Data type:", X.dtype)

# np.array(image)

# image
```

```python
X_temp , X_test , y_temp , y_test = train_test_split(
    X , y,
    test_size=0.20,
    random_state=42,
    stratify = y
)

X_train , X_val , y_train , y_val = train_test_split(
    X_temp,
    y_temp,
    test_size=0.20,
    random_state=42,
    stratify=y_temp
)
```

```python
output_path = Path("../data/processed")
output_path.mkdir(parents=True, exist_ok=True)

np.savez_compressed(
    output_path / "brain_mri_data.npz",
    X_train=X_train,
    X_val=X_val,
    X_test=X_test,
    y_train=y_train,
    y_val=y_val,
    y_test=y_test,
)
```
